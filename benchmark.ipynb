```python
def get_test_case(str):
    if 'grosser_stern' in str:
        return 'grosser_stern'
    elif 'large_street_network' in str:
        return 'large_street_network'
    else:
        return 'unknown'
```

```python
def get_extend(str):
    test_case = get_test_case(str)

    try:
        extend = float(str.split('/')[-1])
        return extend**2
    except Exception:
        if test_case == 'grosser_stern!':
            return 170*135
        elif test_case == 'large_street_network!':
            return 450*607
        else:
            return -1
```

```python
def get_function(str):
    if 'IntersectsAnyVector' in str:
        return 'vector'
    elif 'IntersectsAnySet' in str:
        return 'set'
    else:
        return 'current'
```

```python
df_result = pd.read_csv(StringIO(result), sep=r"\s+", header=None, skipinitialspace=True, skip_blank_lines=True)
df_result = df_result.rename(columns={1: "time", 3: "cpu", 0: 'foo'})
df_result = df_result.drop(columns=[2, 4, 5])

df_result['time'] = df_result['time'] .values / 1000.
df_result['cpu'] = df_result['time'] .values / 1000.
df_result['test'] = df_result.foo.apply(lambda x: get_test_case(x))
df_result['extend'] = df_result.foo.apply(lambda x: get_extend(x))
df_result['function'] = df_result.foo.apply(lambda x: get_function(x))
df_result = df_result.drop(columns='foo')
df_result = df_result[df_result.extend > 0]
df_result = df_result[['test', 'function', 'extend', 'time', 'cpu']]
df_result
```
